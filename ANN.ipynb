```python
data.loc[data['species']=='setosa','species']=0
data.loc[data['species']=='versicolor','species']=1
data.loc[data['species']=='virginica','species']=2
print(data.head())
```

```python
#shiffling data because our data has 150 rows in that first 50 species are  setosa and next 50 versicolor and last 50 are virginica is not good to train the model
data=data.loc[np.random.permutation(len(data))]
data.head()
```

```python
x=data.drop(columns=['species'])
y=data['species']
print(x.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42)
print(X_train.shape)
print(X_test.shape)
```

```python
y_train
```

```python
#neural network modules
from keras.models import Sequential
from keras.layers import Dense, Activation
#from keras.utilts import np_utilts
import tensorflow as tf
```

```python
#change label to one hot encoder
'''
[0]->[1,0,0]
[1]->[0,1,0]
[2]->[0,0,1]
'''
y_train=tf.keras.utils.to_categorical(y_train,num_classes=3)
y_test=tf.keras.utils.to_categorical(y_test,num_classes=3)
print("shape of y_train",y_train.shape)
print("shape of y_test",y_test.shape)
```

```python
#multi-class classfication
#define keras model

model=Sequential()

model.add(Dense(32, input_dim=4, activation='relu'))# 1 hidden layers and input_dim represent how many neurons in input layer
model.add(Dense(64,  activation='relu'))# 2 hidden layer
model.add(Dense(128, activation='relu'))# 3 hidden layer

model.add(Dense(3, activation='softmax'))# output layer
```

```python
model.compile(
    loss='categorical_crossentropy', #error
    optimizer='adam', #gradient decent
    metrics=['accuracy'] #accuracy
)
```

```python
#training the data
model.fit(X_train,y_train,
    validation_data=(X_test, y_test),
    batch_size=30,
    epochs=30)
```

```python
y_pred = model.predict(X_test)
print("accuracy",y_pred)
```
